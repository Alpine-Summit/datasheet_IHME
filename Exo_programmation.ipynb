```python
def moyenne_n(n):
    if n==0:
        print ("Erreur : n doit être supérieur à 0")
        return None
    somme = 0
    for i in range(1, n+1):
        somme = somme + i
    print ("La somme des",n,"premier nombres est", somme)
    moyenne = somme/n
    print ("La moyenne des",n,"premier nombres est", moyenne)
    return moyenne

def somme_n_rec(n):
    if n == 0:
        return 0
    return somme_n_rec(n - 1) + n


def moyenne_n_rec(n):
    if n == 0:
        print("Erreur : n doit être supérieur à 0")
        return None

    somme = somme_n_rec(n)
    moyenne = somme / n

    print("La somme récursive des", n, "premiers nombres est", somme)
    print("La moyenne des", n, "premiers nombres est", moyenne)    
```

```python
moyenne_n(4)
moyenne_n_rec(4)
```

```python
def fact_n(n):
    if n == 0:
        return 1
    fact = 1
    for i in range(1, n + 1):
        fact = fact * i
    print("Le factoriel de", n, "est", fact)
    return fact


def fact_r(n):
    if n == 0:
        return 1
    return fact_r(n - 1) * n


def fact_rec(n):
    fact = fact_r(n)
    print("Le factoriel récursif de", n, "est", fact)
```

```python
fact_n(5)
fact_rec(5)
```

```python
def Fib_n(n):
    if n == 0:
        return 0
    
    a = 0
    b = 1
    
    for i in range(1, n):
        a, b = b, a + b
    
    return b


def Fib_rec(n):
    if n == 0:
        return 0
    if n == 1:
        return 1
    
    return Fib_rec(n - 1) + Fib_rec(n - 2)


memo = {0: 0, 1: 1}

def Fib_rec_memo(n):
    if n in memo:
        return memo[n]
    
    memo[n] = Fib_rec_memo(n - 1) + Fib_rec_memo(n - 2)
    return memo[n]
```

```python
print("Le terme", 10, "de Fibonacci est", Fib_n(10))
print("Le terme récursif", 10, "de Fibonacci est", Fib_rec(10))
print("Le terme avec mémo", 10, "de Fibonacci est", Fib_rec_memo(10))
```

```python
def PlusBase2(x,y):
    z = x + y
    Reste = z-((x+y)//2)*2
    if Reste == 0:
        Reste = "pair"
    else :
        Reste = "impair"
    print("La somme de",x,"et",y,"est égale à",z,"et est",Reste)

def PlusBase2_V2(x,y):
    z = x + y
    Reste = z%2
    if Reste == 0:
        Reste = "pair"
    else :
        Reste = "impair"
    print("La somme de",x,"et",y,"est égale à",z,"et est",Reste)
```

```python
PlusBase2(5,3)
PlusBase2_V2(5,3)
```

```python
# Fonction qui vérifie si un nombre est divisble par 3

def Divisible3(n):
    Reste = n -((n//3))*3
    if Reste == 0:
        Resultat = "est divisible par 3"
    else :
        Resultat = "n'est pas divisible par 3"
    print("Le paramètre n =",n,Resultat,"(reste=",Reste,")")

def Divisible3_V2(n):
    Reste = n%3
    if Reste == 0:
        Resultat = "est divisible par 3"
    else :
        Resultat = "n'est pas divisible par 3"
    print("Le paramètre n =",n,Resultat,"(reste=",Reste,")")
```

```python
Divisible3(8)
Divisible3_V2(8)
```
